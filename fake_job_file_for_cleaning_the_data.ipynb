```python
df.drop(columns=["job_id","department","salary_range"], inplace=True)
```

```python
df.columns
```

```python
categorical_features = df.select_dtypes(include='object').columns
numerical_features = df.select_dtypes(include='number').columns
```

```python
[[categorical_features,
numerical_features]]
```

```python
TEXT_COLUMNS = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

for col in TEXT_COLUMNS:
    df[col] = df[col].fillna("")
```

```python
CATEGORICAL_COLUMNS = [
    "location",
    "employment_type",
    "required_experience",
    "required_education",
    "industry",
    "function"
]

for col in CATEGORICAL_COLUMNS:
    df[col] = df[col].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
# Remove duplicate rows

print("Before:", len(df))

df = df.drop_duplicates().reset_index(drop=True)

print("After:", len(df))
```

```python
# Remove leading/trailing spaces

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
# Replace multiple spaces
# like : Data     Scientist     ----> Data Scientist

import re

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].apply(
        lambda x: re.sub(r"\s+", " ", x) if isinstance(x, str) else x
    )
```

```python
# Remove HTML tags
from bs4 import BeautifulSoup

def remove_html(text):
    if isinstance(text, str):
        return BeautifulSoup(text, "html.parser").get_text()
    return text

for col in TEXT_COLUMNS:
    df[col] = df[col].apply(remove_html)
```

```python
# Remove URLs
import re

def remove_urls(text):
    if isinstance(text, str):
        return re.sub(r"http\S+|www\S+", "", text)
    return text

for col in TEXT_COLUMNS:
    df[col] = df[col].apply(remove_urls)
```

```python
# Remove email addresses
def remove_email(text):
    if isinstance(text, str):
        return re.sub(r'\S+@\S+', '', text)
    return text

for col in TEXT_COLUMNS:
    df[col] = df[col].apply(remove_email)
```

```python
# Remove strange Unicode characters

def clean_unicode(text):
    if isinstance(text, str):
        return text.encode("ascii", "ignore").decode()
    return text

for col in TEXT_COLUMNS:
    df[col] = df[col].apply(clean_unicode)
```

```python
df["text"] = (
    "Title: " + df["title"] +
    " Company Profile: " + df["company_profile"] +
    " Description: " + df["description"] +
    " Requirements: " + df["requirements"] +
    " Benefits: " + df["benefits"] +
    " Industry: " + df["industry"] +
    " Function: " + df["function"] +
    " Employment Type: " + df["employment_type"] +
    " Experience: " + df["required_experience"] +
    " Education: " + df["required_education"] +
    " Location: " + df["location"]
)
```
